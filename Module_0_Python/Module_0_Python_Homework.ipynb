## 📌 Görev 2: Öğrenci Dersi Geçti Mi?
Sınıf geçme notumuz **60** olarak belirlendi.
Sizden `gecti_mi` adında bir fonksiyon yazmanızı istiyoruz. Bu fonksiyon `not_degeri` isminde bir sayı alacak:
- Eğer sayı 60'a eşit veya büyükse `True` döndürecek (return True).
- Eğer sayı 60'tan küçükse `False` döndürecek (return False).

> 💡 **İpucu:** `if not_degeri >= 60:` şeklinde bir şart bloğu kullanabilirsiniz.

---

**English**

## Task 2: Did the student pass?
Write `gecti_mi(not_degeri)`. Return the boolean `True` for grades greater than or equal to 60, and `False` below 60. Hint: use `if not_degeri >= 60:`.

```python
def gecti_mi(not_degeri):
   if not_degeri >= 60:
    return True
   else:
    return False

# Yazdığınız fonksiyonu test edelim: / Test your function:
print("70 Alan biri: / A student with grade 70:", gecti_mi(70)) # True çıkmalı / Expected True
print("45 Alan biri: / A student with grade 45:", gecti_mi(45)) # False çıkmalı / Expected False
```

## 📌 Görev 3: Sadece Geçenleri Ayıralım
Şimdi tüm sınıfın notlarını tek tek inceleyelim ve sadece **dersten geçen** (yani 60 ve üzeri alan) notları yeni bir listeye ekleyelim.

Sizden istenen:
- `gecen_notlar` isimli boş listeyi kullanın.
- Orijinal `notlar` listesi üzerinde bir `for` döngüsü kurun.
- Her bir notu, az önce yazdığınız `gecti_mi` fonksiyonuna (veya direkt `if` şartına) sokun.
- Şartı sağlayanları `.append()` metodu ile listeye ekleyin.

---

**English**

## Task 3: Filter passing grades
Create `gecen_notlar`. Iterate through `notlar` with a for loop, check `gecti_mi` (or use an equivalent if condition), and append grades of 60 or higher in their original order.

```python
gecen_notlar = []

for not_degeri in notlar:
  if gecti_mi(not_degeri):
    gecen_notlar.append(not_degeri)

print("Geçen öğrencilerin notları: / Passing grades:", gecen_notlar)
```

## 📌 Görev 4: Notları İsimlerle Eşleştirelim (Dictionary)
Listeler tek başlarına anlam ifade etmez. Hangi notun kime ait olduğunu bilmek istiyoruz!

Sizden istenen:
- `isimler` listesi ile `notlar` listesini birleştirerek bir Sözlük (Dictionary) oluşturmanız.
- Sözlüğün **Anahtarı (Key)** öğrencinin adı, **Değeri (Value)** ise aldığı not olmalı.
- Sonucu `kare_sozlugu` isimli değişkene atamalısınız.

> 💡 **İpucu:** `zip(isimler, notlar)` kullanarak iki listeyi aynı anda bir `for` döngüsünde dolaşabilirsiniz. Veya boş sözlüğe `kare_sozlugu[isim] = not` şeklinde atama yapabilirsiniz.

Sözlük sekiz isim-not çiftinin tamamını içermelidir.

---

**English**

## Task 4: Match names to grades
Create `kare_sozlugu` from `isimler` and `notlar`. Each key is a student’s name; its value is their grade. Include all eight pairs. Hint: `zip(isimler, notlar)` visits both lists together; assign `kare_sozlugu[isim] = not_degeri`.

```python
isimler = ["Ali", "Ayşe", "Fatma", "Veli", "Ahmet", "Zeynep", "Can", "Elif"]
notlar = [75, 42, 90, 55, 60, 85, 30, 100]

kare_sozlugu = {}

for isim, not_degeri in zip(isimler, notlar):
  kare_sozlugu[isim] = not_degeri

print(kare_sozlugu)
```
